```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])
#Converts a multidimensional array into a one-dimensional array.

b = a.flatten()

print(b)
#flatten() creates a copy. Changes to b do not affect a
```

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = a.ravel()

print(b)
#ravel() tries to return a view of the original array when possible, so changes may affect a. If a view isn't possible, it returns a copy.
```
